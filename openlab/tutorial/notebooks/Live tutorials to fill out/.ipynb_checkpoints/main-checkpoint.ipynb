## HTTP 
Hypertext Transfer Protocol, or HTTP, is the foundation application protocol for the internet (source - wikipedia). The 4 main types that API's use, including Openlab are:
    Post
    Put
    Get
    Delete

The OpenLab python client uses a library (oauth) to handle the dirty work. Each instance of openlab.http_client() creates a client from this library, and all HTTP methods can be called (like the following)


def whoami(self):

        """
        Returns information about the current user
        """
        r=self.client.get(self.url+"/users/whoami")
        return self.standard_response(r, 200)
